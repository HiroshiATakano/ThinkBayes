犯行現場に2人の血痕が残されていた。容疑者オリバーは、採血の結果、血液型はO型だった。2人の血痕はO型（その地域で多い血液型、頻度６０％）とAB型（頻度１％）と判明した。このデータは、オリバーがそのうちの1人であるという命題を裏付ける証拠となるか。

```python
from fractions import Fraction

# 事前オッズ
prior_odds = Fraction(1, 1)

# 血液型頻度
O = Fraction(60, 100)
AB = Fraction(1, 100)

# H: オリバーが2人のうち1人
# その場合、もう1人がAB型である確率
likelihood_H = AB

# not H: オリバーではない2人の場合
# O型とAB型の組み合わせ
likelihood_not_H = 2 * O * AB

# ベイズファクター
bayes_factor = likelihood_H / likelihood_not_H

# 事後オッズ
posterior_odds = prior_odds * bayes_factor

# オッズ → 確率
posterior_prob = posterior_odds / (posterior_odds + 1)


print("尤度")
print(f"P(E|H)       = {likelihood_H}")

print(f"P(E|not H)   = {likelihood_not_H}")

print("\nベイズファクター")
print(f"{likelihood_H} / {likelihood_not_H} = {bayes_factor}")

print("\n事後オッズ")
print(f"{posterior_odds}:1")

print("\n事後確率")
print(f"{posterior_prob}")

print(f"\n小数表示 = {float(posterior_prob):.1%}")
```
