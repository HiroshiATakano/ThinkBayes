```python
from thinkbayes import Suite

class Euro(Suite):

    def Likelihood(self, data, hypo):
        x = hypo
        if data == 'H':
            return x/100.0
        else:
            return 1 - x/100.0
```

```python
posts = range(1,101)
suite = Euro(posts)

for hypo, prob in list(suite.Items())[:5]:
    print(hypo, prob)

dataset = 'H' * 140 + 'T' * 110
```

```python
for data in dataset:
    suite.Update(data)

suite.Items()

post = [v for k, v in suite.Items()]

print(suite.Mean())

```

```python
max(suite.Items(), key=lambda x: x[1])
```

```python
best_hypo, best_prob = max(suite.Items(),
                           key=lambda x: x[1])

print(best_hypo, best_prob)
```

```python
import matplotlib.pyplot as plt
plt.rcParams['font.family'] = 'YuGothic'

hypos = range(1,101)

plt.plot(hypos, post, label="一様事前確率")

plt.xlabel('x')
plt.ylabel('確率')

plt.title('一様事前分布でのユーロ硬貨問題に対する事後確率')
plt.legend()
plt.grid(True)
```
