```python
from thinkbayes import Suite

class Train(Suite):

    def Likelihood(self, data, hypo):
        if hypo < data:
            return 0
        else:
            return 1.0/hypo   
```

```python
for r in [500, 1000, 2000]:
    hypos = range(1,r+1)
    suite = Train(hypos)

    for data in [60,30,90]:
        suite.Update(data)

    print(f'{r} : {suite.Mean():.0f}')
```
