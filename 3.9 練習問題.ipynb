* 通常の事前分布

    * 小さな会社も大きな会社も同じように考える

* サイズバイアス事前分布

    * 大きな会社ほど機関車を見かける確率が高い
    

```python
from thinkbayes import Suite, Pmf


class Train3(Suite):

    def __init__(self, hypos, alpha=1.0):
        Pmf.__init__(self)

        for hypo in hypos:
            self.Set(hypo, hypo**(1-alpha))

        self.Normalize()

    def Likelihood(self, data, hypo):
        if hypo < data:
            return 0
        else:
            return 1.0 / hypo
```

元の事前分布は、

$ P(N) \propto N^{-1} $

サイズバイアス後は、

$ P^{*}(N) \propto N \times N^{-1} = 1 $
