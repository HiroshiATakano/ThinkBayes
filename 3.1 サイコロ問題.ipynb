```python
from thinkbayes import Suite

class Dice(Suite):

    def Likelihood(self, data, hypo):
        if hypo < data:
            return 0
        else:
            return 1.0/hypo
```

```python
suite = Dice([4,6,8,12,20])
suite.Update(6)

suite.Items()
```

```python
for roll in [6,8,7,7,5,4]:
    suite.Update(roll)

suite.Items()
```
