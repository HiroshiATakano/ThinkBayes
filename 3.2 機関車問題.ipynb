```python
from thinkbayes import Suite
import matplotlib.pyplot as plt

class Train(Suite):
    def Likelihood(self, data, hypo):
        if hypo < data:
            return 0
        else:
            return 1.0/hypo        
```

```python
hypos = range(1,1001)
suite = Train(hypos)

suite.Update(60)

print(suite.Mean())
```

```python
post = [v for k, v in suite.Items()]
```

```python
plt.plot(hypos, post)
plt.xlabel("Number of locomotives (N)")
plt.ylabel("Probability")
plt.title("Posterior Distribution")
plt.grid(alpha=0.3)
```
