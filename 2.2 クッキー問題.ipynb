```python
from thinkbayes import Pmf

# prior distribution
pmf = Pmf()
pmf.Set('Bowl1', 0.5)
pmf.Set('Bowl2', 0.5)

# 事前確率x尤度
pmf.Mult('Bowl1', 0.75)
pmf.Mult('Bowl2', 0.5)

# 再正規化 (renormalize)
pmf.Normalize()

print(f'Bowl1 : {pmf.Prob('Bowl1'):.1f}', f'Bowl2 : {pmf.Prob('Bowl2'):.1f}')

for hypo, prob in pmf.Items():
    print(f'{hypo}, {prob:.1f}' )
```

```python
from thinkbayes import Suite

class Cookie2(Suite):
    mixes = {
        'Bowl1':dict(vanilla=0.75, chocolate=0.25),
        'Bowl2':dict(vanilla=0.5, chocolate=0.5)
    }

    def Likelihood(self, data, hypo):
        print(data)
        mix = self.mixes[hypo]
        print(mix)
        like = mix[data]
        return like
```

```python
hypo = ['Bowl1','Bowl2']

suite = Cookie2(hypo)
suite.Update('vanilla')
suite.Update('chocolate')

suite.Items()


for hypo, prob in suite.Items():
    print(f'{hypo}, {prob:.2f}' )

```
