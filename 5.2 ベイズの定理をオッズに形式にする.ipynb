事後オッズ (posterior odds) は、事前オッズ (prior odds) に尤度の比ををかけたものになる



クッキー問題を、オッズで解くPythonプログラムを作成して。「クッキーの入ったボウルが二つある。ボウル１には30枚のバニラクッキーと10枚のチョコレートクッキがある。ボウル２にはそれぞれ２０枚のバニラとチョコレートクッキーがある。どちらかのボウルをランダムに選び、目をつぶってクッキーを一枚取り出したら。バニラだった。その時それがボウル１のだという確率はいくらか。

```python
# Cookie Problem (Odds Version)

prior_odds = 1.0

# バニラが出る確率（尤度）
like_bowl1 = 30/40
like_bowl2 = 20/40

# 尤度比（ベイズファクター_）
bayes_factor = like_bowl1/like_bowl2

# 事後オッズ
posterior_odds = prior_odds * bayes_factor

# 事後確率
posterior_prob = posterior_odds / (1 + posterior_odds)

print(f"事前オッズ: {posterior_odds:.2f} : 1")
print(f"尤度比： {bayes_factor:.2f}")
print(f"事後オッズ： {posterior_odds:.2f} : 1")
print(f"ボウル１の確率： {posterior_prob:.1%}")
```

```python
from fractions import Fraction

# 事前オッズ
prior_odds = Fraction(1, 1)

# バニラの尤度
like_bowl1 = Fraction(30, 40)
like_bowl2 = Fraction(20, 40)

# ベイズファクター（尤度比）
bayes_factor = like_bowl1 / like_bowl2

# 事後オッズ
posterior_odds = prior_odds * bayes_factor

# オッズ → 確率
posterior_prob = posterior_odds / (posterior_odds + 1)

print("事前オッズ")
print(f"  {prior_odds}")

print("\n尤度")
print(f"  ボウル1 : {like_bowl1}")
print(f"  ボウル2 : {like_bowl2}")

print("\nベイズファクター")
print(f"  {like_bowl1} ÷ {like_bowl2} = {bayes_factor}")

print("\n事後オッズ")
print(f"  {prior_odds} × {bayes_factor} = {posterior_odds}")

print("\n事後確率")
print(f"  {posterior_odds} / ({posterior_odds} + 1) = {posterior_prob}")

print(f"\n小数表示 : {float(posterior_prob):.1%}")
```
