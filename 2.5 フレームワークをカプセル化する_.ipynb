```python
from thinkbayes import Pmf

class Suite(Pmf):
    def __init__(self, hypo=tuple()):
        """分布を初期化する"""

    def Update(self, data):
        """仮説に基づいて各仮説を更新する"""

    def Print(self):
        """仮説とその確率を出力する"""
```

```python
from thinkbayes import Suite

class Monty(Suite):
    def Likelihood(sefl, data, hypo):
        if hypo == data:
            return 0
        elif hypo == 'A':
            return 0.5
        else:
            return 1
```

```python
suite = Monty("ABC")
suite.Update("B")
suite.Items()
```
