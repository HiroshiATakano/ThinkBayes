```python
from thinkbayes import Suite

class Euro(Suite):

    def Likelihood(self, data, hypo):
        x = hypo
        if data == 'H':
            return x/100.0
        else:
            return 1 - x/100.0
        
def TrianglePrior():
    suite = Euro(name='triangle')
    for x in range(0,51):
        suite.Set(x,x)
    for x in range(51,101):
        suite.Set(x,100-x)
    suite.Normalize()

    return suite

    
suite = TrianglePrior()

d = [k for v, k in suite.Items()]

suite.name
```

```python


x = range(1,len(suite)+1)

plt.plot(x, d, label="三角事前確率")

plt.xlabel('x')
plt.ylabel('確率')

plt.title('三角事前確率')
plt.legend()
plt.grid(True)
```

```python

posts = range(1,101)
suite = Euro(posts)
dataset = 'H' * 140 + 'T' * 110

for data in dataset:
    suite.Update(data)

post = [v for k, v in suite.Items()]

print(suite.Mean())

hypos = range(1,101)

plt.plot(hypos, post, label="三角事前確率")

plt.xlabel('x')
plt.ylabel('確率')

plt.title('一様事前分布でのユーロ硬貨問題に対する事後確率')
plt.legend()
plt.grid(True)
```
