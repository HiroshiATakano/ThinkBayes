```python
from thinkbayes import Suite

class TraceDice(Suite):

    def __init__(self, hypos):
        super().__init__()
        for h in hypos:
            self.Set(h, 1)

        print("0 prior")
        print(self.d)

    # ★これが必須（重要）
    def Likelihood(self, data, hypo):
        if hypo < data:
            return 0
        return 1 / hypo

    def Update(self, data):

        print("\n=== UPDATE START ===")

        for h in self.Values():
            like = self.Likelihood(data, h)
            self.Mult(h, like)
            print(f"after Mult({h})")
            print(self.d)

        print("\nbefore Normalize")
        print(self.d)

        self.Normalize()

        print("\nafter Normalize (posterior)")
        print(self.d)
```

```python
hypos = [4, 6, 8, 12, 20]

dice = TraceDice(hypos)

dice.Update(11)
dice.Update(5)
dice.Update(9)
```

```python
from thinkbayes import Pmf

# -----------------------------
# 1. prior（Setの状態）
# -----------------------------
pmf = Pmf()

hypos = ['A', 'B', 'C']

for h in hypos:
    pmf.Set(h, 1)

print("0: prior (Set後)")
print(pmf.Items())

# -----------------------------
# 2. likelihood（Multの中身）
# -----------------------------
def likelihood(h):
    if h == 'A':
        return 1
    if h == 'B':
        return 5
    if h == 'C':
        return 1

# -----------------------------
# 3. Mult（未正規化posterior）
# -----------------------------
for h in pmf.Values():
    pmf.Mult(h, likelihood(h))
    print(f"after Mult {h}: {pmf.Items()}")

print("\n1: after likelihood (unnormalized)")
print(pmf.Items())

# -----------------------------
# 4. Normalize（確率化）
# -----------------------------
pmf.Normalize()

print("\n2: posterior (normalized)")
print(pmf.Items())
```
