```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.stats import norm

import gspread 
from google.oauth2.service_account import Credentials
```

```python
from googleapiclient.discovery import build

SERVICE_ACCOUNT_FILE = "my-project-vscode-452201-249bdb033f8c.json"  # JSONファイルのパス
SCOPES = ["https://www.googleapis.com/auth/drive"]

credentials = Credentials.from_service_account_file(SERVICE_ACCOUNT_FILE, scopes=SCOPES)
client = gspread.authorize(credentials)

service = build('drive','v3', credentials=credentials)

result = service.files().list(
    includeItemsFromAllDrives=True,
    supportsAllDrives=True
).execute()

print(result)
```

```python
def extract_folder_id(url: str) -> str:
    # "folders/" の後ろを取る
    part = url.split("folders/")[1]
    # "?usp=..." など余計なパラメータがあれば切り落とす
    folder_id = part.split("?")[0]
    return folder_id

# 例
url = "https://drive.google.com/drive/folders/1AbCdEfGhIjKlMnOpQrStUvWxYz?usp=drive_link"
print(extract_folder_id(url))
```
