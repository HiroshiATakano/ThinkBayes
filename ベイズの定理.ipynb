# ThinkBayes

## 1.1 条件付き確率

#### Bが真の場合のAの確率

$ P(A|B) $

## 1.2 結合確率

#### 2つの事象が独立でない場合に、同時確率（AかつB）を求める

Aの確率 x Aだった場合にBが起きる確率

$ P(AかつB) = p(A)p(B|A) $

## 1.3 クッキー問題

クッキーが入った２つのボウルがある

ボウル１には、30枚のバニラクッキー、10枚のチョコレートクッキーがある

ボウル２には、それぞれ20枚ずつのクッキーがある

どちらかのボウルをランダムに選び、目をつぶってクッキーを一つランダムに選ぶ。

そのクッキーがバニラだった場合に、それがボウル１のだという確率はいくらだろう

## ベイズの定理

$ p(B)p(A|B) = p(A)p(B|A) $

$ p(A|B) = \frac{p(A)p(B|A)}{p(B)} $

### クッキー問題

$ p(B_1|V) = \frac{p(B_1)p(V|B_1)}{p(V)} $

クッキーがバニラだった時に、ボウル１である確率を求める。

$ p(B_1|V) = $

$ p(B_1) = 1/2 $ ボウル１を選ぶ確率は 1/2

$ p(V|B_1) = 3/4 $ ボウル１を選んだ時のバニラクッキーの出る確率

$ p(V) = 5/8 $ どちらかのボウルからバニラクッキーをとる確率、両方のボウルに５０枚のバニラクッキーと、30枚のチョコレートクッキーがある

### 取り出したクッキーは食べてしまい補充しない

観測を重ねるたびにボウルの中身が変化する。

逐次更新 (Sequential Updating)

どちらかのボウルをランダムに選ぶ。

その後、クッキーを２回取り出す。取り出したクッキーは食べてしまいボウルに戻さない。

#### 仮説：

H1 ボウル１を選んだ

H2 ボウル２を選んだ

#### 事前確率

$ p(H_1) = p(H_2) = 0.5 $

#### 尤度



```python
# =====================================
# クッキー問題
# バニラ → チョコ
# 取り出したら戻さない
# =====================================

# ----------------------------
# 初期状態
# ----------------------------

p_bowl1 = 0.5
p_bowl2 = 0.5

print("初期状態")
print(f"ボウル1: {p_bowl1:.4f}")
print(f"ボウル2: {p_bowl2:.4f}")

# =====================================
# 1回目：バニラ
# =====================================

like_b1 = 30 / 40
like_b2 = 20 / 40

p_bowl1 *= like_b1
p_bowl2 *= like_b2

norm = p_bowl1 + p_bowl2

p_bowl1 /= norm
p_bowl2 /= norm

print("\n1回目：バニラ")

print(f"ボウル1: {p_bowl1:.4f}")
print(f"ボウル2: {p_bowl2:.4f}")

# =====================================
# 2回目：チョコ
# =====================================

# バニラを食べた後

like_b1 = 10 / 39
like_b2 = 20 / 39

p_bowl1 *= like_b1
p_bowl2 *= like_b2

norm = p_bowl1 + p_bowl2

p_bowl1 /= norm
p_bowl2 /= norm

print("\n2回目：チョコ")

print(f"ボウル1: {p_bowl1:.4f}")
print(f"ボウル2: {p_bowl2:.4f}")
```

## 1.5 通時的解釈　(diachroic interpretaion)

$ p(H|D) = \frac{p(H)p(D|H)}{p(D)} $

$ p(H) $ データを見る前の仮説の確率を表し、事前確率 (prior probability)

$ p(H|D) $ 求めたい、データを見た後の仮説の確率、事後確率 (posterior probability)

$ p(D|H) $ 仮説のもとでのデータの確率で、尤度 (likelihood)

$ p(D) $ どのような仮説であってもデータの得られる確率で、正規か定数 (normalization constant)



## 1.6 M&M's問題

1995年以前の色の組み合わせ：茶３０％、黄２０％、赤２０％、緑１０％、橙１０％、黄褐色１０％
1995年以降の色の組み合わせ：青２４％、緑２０％、橙１４％、黄１４％、赤１３＃、茶１３％

友達が、１９９４年と１９９６年の袋から一つづつM&M'sをくれた、一つが黄色で、もう一つは緑だった。黄色が１９９４年の袋からだという確率はいくらか。



### 1. 仮説 (Hypothesis)

仮説 $ H_1 $ 
* 黄色 -> 1994年
* 緑 -> 1996年

仮説 $ H_2 $
* 黄色 -> 1996年
* 緑 -> 1994年

### 2. 事前確率 (Prior)

$ p(H_1) = 0.5 $

$ p(H_2) = 0.5 $

### 3. 尤度 (Likelihood)

$ H_1 $ の場合

1994年の袋の黄色の確率: 0.20

1996年の袋の緑の確率： 0.20

$ p(D|H_1) = 0.2 \times 0.2 = 0.04 $

$ H_2 $ の場合

1996年の袋の黄色の確率: 0.14

1994年の袋の緑の確率： 0.10

$ p(D|H_2) = 0.14 \times 0.1 = 0.014 $

### 4. ベイズ更新（事後確率）

事後確率 $ \propto $ 事前確率　x 尤度

H1: 0.5 x 0.04 = 0.02

H2: 0.5 x 0.014 = 0.007

### 5. 正規化　（合計を１にする）

$ 0.02 + 0.007 = 0.027 $

$ p(H_1|D) = 0.02 / 0.027 = 0.7407 $

### 6. 答え

74.1 %

```python
# M&M's 問題
# ベイズ推論の例

# -----------------------------
# 1994年袋（旧配色）
# -----------------------------
bag_1994 = {
    "brown": 0.30,
    "yellow": 0.20,
    "red": 0.20,
    "green": 0.10,
    "orange": 0.10,
    "tan": 0.10
}

# -----------------------------
# 1996年袋（新配色）
# -----------------------------
bag_1996 = {
    "blue": 0.24,
    "green": 0.20,
    "orange": 0.14,
    "yellow": 0.14,
    "red": 0.13,
    "brown": 0.13
}

# ====================================================
# 仮説 H1:
# 黄色 → 1994年
# 緑   → 1996年
# ====================================================

likelihood_H1 = (
    bag_1994["yellow"] *
    bag_1996["green"]
)

# ====================================================
# 仮説 H2:
# 黄色 → 1996年
# 緑   → 1994年
# ====================================================

likelihood_H2 = (
    bag_1996["yellow"] *
    bag_1994["green"]
)

# -----------------------------
# 事前確率
# -----------------------------
prior_H1 = 0.5
prior_H2 = 0.5

# -----------------------------
# 非正規化事後確率
# -----------------------------
posterior_H1 = prior_H1 * likelihood_H1
posterior_H2 = prior_H2 * likelihood_H2

# -----------------------------
# 正規化
# -----------------------------
total = posterior_H1 + posterior_H2

posterior_H1 /= total
posterior_H2 /= total

# -----------------------------
# 結果表示
# -----------------------------
print("H1 の尤度 =", likelihood_H1)
print("H2 の尤度 =", likelihood_H2)

print()

print("黄色が1994年袋から来た確率")
print(f"{posterior_H1:.4f}")

print()

print("％表示")
print(f"{posterior_H1 * 100:.2f}%")
```

### ガチャガム問題

ガムメーカーが2020年に配合を変更した。

友達が2019年の袋から１個、2021年の袋から１個ガムを取り出したた。色は赤、緑でした。

赤が2019年の袋から出た確率は？

```python
# ==================================
# ガチャガム問題
# ==================================

bag_2019 = {
    "red": 0.40,
    "blue": 0.30,
    "yellow": 0.20,
    "green": 0.10
}

bag_2021 = {
    "red": 0.20,
    "blue": 0.40,
    "yellow": 0.10,
    "green": 0.30
}

# 仮説 H1
# 赤ー＞2019
# 緑ー＞2021

likelihood_H1 = (
    bag_2019["red"] * bag_2021["green"]
)

# 仮説 H1
# 赤ー＞2021
# 緑ー＞2019

likelihood_H2 = (
    bag_2021["red"] * bag_2019["green"]
)

# 事前確率

prior_H1 = 0.5
prior_H2 = 0.5

# 非正規化事後確率

posterior_H1 = prior_H1 * likelihood_H1
posterior_H2 = prior_H2 * likelihood_H2

# 正規化

total = posterior_H1 + posterior_H2

posterior_H1 /= total
posterior_H2 /= total

# 結果

print("尤度 H1", likelihood_H1)
print("尤度 H2", likelihood_H2)

print()

print(f"赤が2019年の袋から出た確率")
print(f"{posterior_H1:.4f}")
```

３つの工場があります

市場で商品を１個買ったところ、その商品は不良でした。その不良品が工場Cで作られた確率を求める。

工場A

不良率：
$ p(不良率|A) = 0.01 $

生産割合：
40%

工場B

不良率：
$ p(不良率|B) = 0.03 $

生産割合
35%

工場C

不良率：
$ p(不良率|C) = 0.05 $

生産割合
25%

```python
# ===================================
# 3つの仮説のベイズ推論
# ===================================

# 事前確率（生産割合）

priors = {
    "A": 0.40,
    "B": 0.35,
    "C": 0.25
}

# 尤度
# P(不良 | 工場)

likelihoods = {
    "A": 0.01,
    "B": 0.03,
    "C": 0.05
}

# 事前x尤度

unnormalized = {}

for factory in priors:
    unnormalized[factory] = (
        priors[factory] * likelihoods[factory]
    )

# 正規化定数

evidence = sum(unnormalized.values())

# 事後確率

posteriors = {}

for factory in unnormalized:
    posteriors[factory] = (
        unnormalized[factory] / evidence
    )

print("事後確率")

for factory in posteriors:
    print(f"工場{factory}: " f"{posteriors[factory]:.4f}")
```

## 1.7 モンティ・ホール問題

* モンティが３つの扉を示して「扉の後ろに商品があります」と告げる。１つが車で、残りの２つはピーナツバターとかつけ爪など車ほど価値のないものである。商品はランダムに配置される。
* ゲームの目的は、どこに車があるか当てること、当てたら車がもらえる。
* 挑戦者が扉を選ぶ。それを扉Aとする。他は、B、Cとする。
* 挑戦者が選んだ扉を開ける前に、モンティはその場を盛り上げるために、BかCの扉のうち車がない方を開ける。
* そして、挑戦者に「選ぶ扉をそのままにしておいても、変えてもいいですよ」と話す。

問題は、選択した扉を「そのまま」にすべきか、「変える」べきか、また、違いがあるかどうか。



### 挑戦者は扉Aを選んだ

### 事前確率

* 車がAにある確率 $ = 1/3 $
* 車がBにある確率 $ = 1/3 $
* 車がCにある確率 $ = 1/3 $

### 観測データ

Cを開けたらピーナツバターだった

$ D = $ 「モンティがCを開けた

### 仮説を作る

H_A 

車はA： $ p(H_A) = 1/3 $

H_B 

車はB： $ p(H_B) = 1/3 $

H_C 

車はB： $ p(H_C) = 1/3 $

### 尤度を考える

H_Aの場合：車はAで、B、Cも外れ。モンティはBを開けても良いし、Cを開けても良い。ランダムに選ぶとすると。

$ p(D|H_A) = 1/2 $

H_Bの場合：車はBで、Cはハズレ。モンティはCを開ける。

$ p(D|H_B) = 1 $

H_Cの場合：車はCで、モンティはCを開けれらない。

$ p(D|H_C) = 0 $

### ベイズ更新

事前x尤度

A： $ 1/3 \times 1/2 = 1/6 $

B: $ 1/3 \times 1 = 1/3 $

C: $ 1/3 \times 0 = 0 $

合計：　$ 1/6 + 1/3 = 1/2 $

### 正規化

A: 1/3

B: 2/3

C: 0

```python
# ==========================================
# モンティ・ホール問題
# ベイズ推論版
# ==========================================

# 事前確率

priors = {
    "A": 1/3,
    "B": 1/3,
    "C": 1/3
}

# ----------------------------------
# 観測
#
# 「モンティがCを開けた」
# ----------------------------------

likelihoods = {
    "A": 1/2,  # Aが車ならB,Cどちらも開けられる
    "B": 1.0,  # Bが車ならCしか開けられない
    "C": 0.0   # Cが車ならCは開けられない
}

# ----------------------------------
# 非正規化事後確率
# ----------------------------------

unnormalized = {}

for door in priors:
    unnormalized[door] = (
        priors[door]
        * likelihoods[door]
    )

# ----------------------------------
# 証拠 P(D)
# ----------------------------------

evidence = sum(unnormalized.values())

# ----------------------------------
# 事後確率
# ----------------------------------

posteriors = {}

for door in unnormalized:
    posteriors[door] = (
        unnormalized[door]
        / evidence
    )

# ----------------------------------
# 表示
# ----------------------------------

print("事後確率")

for door in posteriors:
    print(
        f"車が扉{door}にある確率 "
        f"{posteriors[door]:.4f}"
    )

print()
print(
    "変更した場合の勝率:",
    f"{posteriors['B']:.4f}"
)
```
